# Python (Plotly) - Testovací cvičení

V tomto sešitu vytváříme interaktivní vizualizace z datasetu `loans_df` pomocí knihovny **Plotly**:
1. Sloupcový graf průměrné výše půjčky podle typu oblasti (`area`) a statusu půjčky (`status` - dvě barvy vedle sebe).
2. Bodový graf (scatter plot) vztahu mezi příjmem žadatele (`applicant_income`) a výší půjčky (`loan_amount`) s barvou bodů podle proměnné `married`.
3. Korelační matice spojitých numerických proměnných s barevnou paletou `Oranges`.

Každý graf obsahuje nadpis i popisky os X a Y.

## 0. Příprava dat (loans_df)

```python
import os
import pandas as pd
import plotly.express as px

data_path = os.path.join("data", "loans.csv") if os.path.exists(os.path.join("data", "loans.csv")) else os.path.join("00_Prework", "data", "loans.csv")
loans_df = pd.read_csv(data_path)
loans_df.columns = loans_df.columns.str.lower()

for col in ['gender', 'married', 'dependents', 'self_employed']:
    loans_df[col] = loans_df[col].fillna(loans_df[col].mode()[0])

loans_df.head(3)
```

## 1. Průměrná výše půjčky podle typu oblasti a statusu půjčky
- Sloupce rozdělené podle `area`
- Dva sloupce vedle sebe odlišené barvou podle `status` (`barmode='group'`)
- Nadpis i popisky os X a Y

```python
avg_loan_by_area_status = loans_df.groupby(['area', 'status'], as_index=False)['loan_amount'].mean()

fig1 = px.bar(
    avg_loan_by_area_status,
    x='area',
    y='loan_amount',
    color='status',
    barmode='group',
    title='Average Property Loan Amount by Area Type and Loan Status',
    labels={
        'area': 'Property Area',
        'loan_amount': 'Average Loan Amount',
        'status': 'Loan Status'
    },
    color_discrete_map={'Y': '#2ca02c', 'N': '#d62728'},
    text_auto='.2s'
)

fig1.update_layout(
    xaxis_title='Property Area',
    yaxis_title='Average Loan Amount',
    legend_title_text='Loan Status',
    template='plotly_white'
)

fig1.show()
```

## 2. Vztah mezi příjmem žadatele a výší půjčky
- Osa X: příjem žadatele (`applicant_income`)
- Osa Y: výše půjčky (`loan_amount`)
- Barva bodů: rodinný stav (`married`)

```python
fig2 = px.scatter(
    loans_df,
    x='applicant_income',
    y='loan_amount',
    color='married',
    title='Relationship Between Borrower Income and Loan Amount by Marital Status',
    labels={
        'applicant_income': 'Borrower Income (Applicant Income)',
        'loan_amount': 'Loan Amount',
        'married': 'Married'
    },
    hover_data=['education', 'self_employed', 'status'],
    opacity=0.7,
    template='plotly_white'
)

fig2.update_layout(
    xaxis_title='Borrower Income (Applicant Income)',
    yaxis_title='Loan Amount',
    legend_title_text='Married'
)

fig2.show()
```

## 3. Korelační matice spojitých numerických proměnných
- Spojité numerické proměnné: `applicant_income`, `coapplicant_income`, `loan_amount`
- Barevná paleta: `Oranges`

```python
continuous_vars = ['applicant_income', 'coapplicant_income', 'loan_amount']
corr_matrix = loans_df[continuous_vars].corr().round(2)

fig3 = px.imshow(
    corr_matrix,
    text_auto=True,
    aspect='auto',
    color_continuous_scale='Oranges',
    title='Correlation Matrix of Continuous Numerical Variables',
    labels=dict(x='Variables', y='Variables', color='Correlation')
)

fig3.update_layout(
    xaxis_title='Continuous Numerical Variables',
    yaxis_title='Continuous Numerical Variables',
    template='plotly_white'
)

fig3.show()
```
