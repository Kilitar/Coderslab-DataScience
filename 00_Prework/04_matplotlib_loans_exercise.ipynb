# Python (Matplotlib) - Testovací cvičení

Vizualizace dat z připraveného datasetu `loans_df` pomocí knihovny **Matplotlib**:
1. Sloupcový graf počtu žadatelů podle pohlaví (`color='lightblue'`, volitelně hodnoty nad sloupci).
2. Sloupcový graf počtu žadatelů podle stavu schválení půjčky (`color='orange'`).
3. Histogram rozdělení proměnné `loan_amount` (50 binů, `color='red'`).

## 0. Příprava dat (loans_df z předchozí úlohy)

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Načtení dat a sjednocení názvů sloupců
data_path = os.path.join("data", "loans.csv") if os.path.exists(os.path.join("data", "loans.csv")) else os.path.join("00_Prework", "data", "loans.csv")
loans_df = pd.read_csv(data_path)
loans_df.columns = loans_df.columns.str.lower()

# Imputace módem u kategoriálních proměnných s chybějícími hodnotami
for col in ['gender', 'married', 'dependents', 'self_employed']:
    loans_df[col] = loans_df[col].fillna(loans_df[col].mode()[0])

loans_df.head(3)
```

## 1. Sloupcový graf počtu žadatelů podle pohlaví
- Barva sloupců: `lightblue`
- Nadpisy grafu a os X a Y
- Hodnoty přímo nad sloupci

```python
plt.figure(figsize=(7, 5))
gender_counts = loans_df['gender'].value_counts()

bars = plt.bar(
    gender_counts.index,
    gender_counts.values,
    color='lightblue',
    edgecolor='steelblue',
    width=0.5
)

plt.title("Number of Borrowers by Gender", fontsize=14, fontweight='bold', pad=12)
plt.xlabel("Gender", fontsize=11, fontweight='semibold')
plt.ylabel("Number of Borrowers", fontsize=11, fontweight='semibold')
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Zobrazení číselných hodnot nad sloupci
plt.bar_label(bars, padding=3, fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 2. Počet žadatelů podle stavu půjčky (Status: Granted / Not Granted)
- Barva sloupců: `orange`
- Nadpisy grafu a obou os

```python
plt.figure(figsize=(7, 5))
status_counts = loans_df['status'].value_counts()
labels = ['Granted (Y)' if s == 'Y' else 'Not Granted (N)' for s in status_counts.index]

bars = plt.bar(
    labels,
    status_counts.values,
    color='orange',
    edgecolor='darkorange',
    width=0.5
)

plt.title("Number of Borrowers by Loan Status", fontsize=14, fontweight='bold', pad=12)
plt.xlabel("Loan Status", fontsize=11, fontweight='semibold')
plt.ylabel("Number of Borrowers", fontsize=11, fontweight='semibold')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.bar_label(bars, padding=3, fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 3. Histogram rozdělení proměnné loan_amount
- Rozdělení do 50 binů
- Barva sloupců: `red`
- Nadpisy grafu a obou os

```python
plt.figure(figsize=(9, 5))

plt.hist(
    loans_df['loan_amount'],
    bins=50,
    color='red',
    edgecolor='darkred',
    alpha=0.75
)

plt.title("Distribution of Loan Amount", fontsize=14, fontweight='bold', pad=12)
plt.xlabel("Loan Amount", fontsize=11, fontweight='semibold')
plt.ylabel("Frequency", fontsize=11, fontweight='semibold')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
