## 1. Připojení Google Drive
Při spuštění v Google Colab budete vyzváni k udělení oprávnění pro přístup k Disku Google.

```
# Připojení Google Drive (funguje v prostředí Google Colab)
try:
    from google.colab import drive
    drive.mount('/content/drive')
    base_dir = '/content/drive/MyDrive/Machine Learning/Google Colab'
    print("Google Drive úspěšně připojen.")
except ImportError:
    # Lokální fallback pro spuštění v lokálním IDE
    import os
    base_dir = os.path.join(os.getcwd(), '00_Prework', 'data') if os.path.exists('00_Prework/data') else 'data'
    print("Spuštěno lokálně, data budou načítána z:", base_dir)
```

## 2. Kontrola verze knihovny Pandas

```
import pandas as pd

print(f"Nainstalovaná verze knihovny Pandas: {pd.__version__}")
```

## 3. Načtení dat ze student_scores.csv
Načteme data ze složky `Machine Learning/Google Colab` na Google Disku do proměnné `df`.

```
import os

input_file = os.path.join(base_dir, 'student_scores.csv')

# Ošetření možné mezery v názvu 'My Drive' vs 'MyDrive'
if not os.path.exists(input_file) and os.path.exists('/content/drive/My Drive/Machine Learning/Google Colab/student_scores.csv'):
    input_file = '/content/drive/My Drive/Machine Learning/Google Colab/student_scores.csv'

df = pd.read_csv(input_file)

# Zobrazení prvních 5 řádků a informací o datasetu
print(f"Rozměry tabulky: {df.shape[0]} řádků, {df.shape[1]} sloupců")
df.head()
```

## 4. Přidání sloupce TotalScore
Sloupec `TotalScore` je součtem sloupců `MathScore`, `ReadingScore` a `WritingScore`.

```
df['TotalScore'] = df['MathScore'] + df['ReadingScore'] + df['WritingScore']

# Kontrola výsledku
df[['MathScore', 'ReadingScore', 'WritingScore', 'TotalScore']].head()
```

## 5. Uložení výsledného DataFrame na Google Disk
Uložíme data do souboru `colab_exercise.csv` na připojený Google Disk.

```
output_file = os.path.join(base_dir, 'colab_exercise.csv')
df.to_csv(output_file, index=False)

print(f"Soubor byl úspěšně uložen do: {output_file}")
```
