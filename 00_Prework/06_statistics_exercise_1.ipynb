# Statistika - Cvičení 1 (NumPy)

V tomto sešitu počítáme základní popisné statistiky pro ceny nemovitostí na Manhattanu v New Yorku pomocí knihovny **NumPy**:
- Extrakce hodnot ze slovníku do NumPy pole (`np.array`)
- Aritmetický průměr (`np.mean`)
- Medián (`np.median`)
- Minimum a maximum (`np.min`, `np.max`)
- Rozptyl (`np.var`)
- Směrodatná odchylka (`np.std`)

## 1. Definice vstupního slovníku a extrakce do NumPy pole

```python
import numpy as np

real_estate_prices = {
    'Broadway': 1200000,
    'Park Avenue': 3500000,
    'Madison Avenue': 2800000,
    'Fifth Avenue': 5000000,
    'Lexington Avenue': 2100000,
    'West End Avenue': 1800000,
    'Central Park West': 4500000,
    'Riverside Drive': 1600000,
    'Waverly Place': 1100000,
    'Bleecker Street': 900000,
    'Christopher Street': 950000,
    'Houston Street': 1300000,
    'Canal Street': 800000,
    'Bowery': 700000,
    'Delancey Street': 1000000
}

# Extrakce cen do NumPy pole
prices = np.array(list(real_estate_prices.values()))
prices
```

## 2. Výpočet statistických ukazatelů a uložení do proměnných

```python
mean_price = np.mean(prices)
median_price = np.median(prices)
min_price = np.min(prices)
max_price = np.max(prices)
var_price = np.var(prices)
std_price = np.std(prices)

print(f"Mean:               ${mean_price:,.2f}")
print(f"Median:             ${median_price:,.2f}")
print(f"Minimum value:      ${min_price:,.2f}")
print(f"Maximum value:      ${max_price:,.2f}")
print(f"Variance:           ${var_price:,.2f}")
print(f"Standard deviation: ${std_price:,.2f}")
```
