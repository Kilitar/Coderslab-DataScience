## Úkol 1: Funkce calculate(a, b, math_operation)
Napište funkci, která přijímá tři argumenty: `a`, `b` a `math_operation` (`add`, `sub`, `mul`, `div`). Pro operace závislé na pořadí je větší číslo na prvním místě. Výsledek je zaokrouhlen na 2 desetinná místa.

```python
def calculate(a: float, b: float, math_operation: str) -> float:
    larger = max(a, b)
    smaller = min(a, b)

    if math_operation == "add":
        result = larger + smaller
    elif math_operation == "sub":
        result = larger - smaller
    elif math_operation == "mul":
        result = larger * smaller
    elif math_operation == "div":
        if smaller == 0:
            raise ZeroDivisionError("Dělení nulou není povoleno.")
        result = larger / smaller
    else:
        raise ValueError(f"Neznámá operace: '{math_operation}'. Povolené jsou: add, sub, mul, div")

    return round(result, 2)

# Testy
print("add (2, 5):", calculate(2, 5, "add"))
print("sub (3, 10):", calculate(3, 10, "sub"))
print("mul (4, 2.5):", calculate(4, 2.5, "mul"))
print("div (3, 10):", calculate(3, 10, "div"))
```

## Úkol 2: Filtrování jmen začínajících na 'C'
Napište funkci, která přijímá seznam křestních jmen a vrací pouze ta, která začínají na písmeno 'C'.

```python
def filter_c_names(names: list[str]) -> list[str]:
    return [name for name in names if name.startswith("C")]

royal_family_names = ["Elizabeth", "Phillip", "Charles", "Camilla", "William", "Catherine", "Anne", "Harry", "Meghan"]
result = filter_c_names(royal_family_names)
print("Jména začínající na 'C':", result)
```

## Úkol 3: Filtrování zaměstnanců podle platu
Funkce vrací slovník zaměstnanců s platem vyšším než zadaná hodnota.

```python
def filter_employees_by_salary(employees: dict, threshold: float) -> dict:
    return {
        name: data
        for name, data in employees.items()
        if data[1] > threshold
    }

company_employees = {
    "Jan Novák": ("Data Analyst", 55000),
    "Petr Svoboda": ("Junior Python Developer", 45000),
    "Eva Černá": ("Senior ML Engineer", 120000),
    "Lucie Bílá": ("Product Owner", 85000),
    "Martin Dvořák": ("QA Specialist", 40000),
    "Karel Procházka": ("Data Scientist", 95000)
}

threshold = 60000
filtered = filter_employees_by_salary(company_employees, threshold)
print(f"Zaměstnanci s platem nad {threshold} Kč:")
for name, (position, salary) in filtered.items():
    print(f" - {name}: {position} ({salary:,} Kč)")
```
