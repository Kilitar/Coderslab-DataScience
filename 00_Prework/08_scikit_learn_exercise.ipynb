# Scikit-learn - Cvičení: Klasifikace a model k-NN

V tomto sešitu si procvičíme základní pracovní postup (pipeline) v knihovně **Scikit-learn**:
1. **Import potřebných modulů a tříd** (`datasets`, `model_selection`, `neighbors`, `metrics`).
2. **Generování syntetického klasifikačního datasetu** pomocí `make_classification`.
3. **Rozdělení dat na trénovací a testovací sadu** (`train_test_split`) v poměru 70:30.
4. **Vytvoření instance klasifikátoru k-nejbližších sousedů** (`KNeighborsClassifier`) s $k = 4$.
5. **Natrénování modelu** na trénovacích datech a predikce na testovacích datech.
6. **Vyhodnocení úspěšnosti modelu** pomocí metriky **Accuracy**.

## 1. Import knihoven a modulů

- `make_classification` z `sklearn.datasets` pro vygenerování syntetického datasetu pro klasifikaci.
- `train_test_split` z `sklearn.model_selection` pro rozdělení dat na trénovací a testovací množinu.
- `KNeighborsClassifier` z `sklearn.neighbors` pro klasifikační algoritmus k-nejbližších sousedů.
- `accuracy_score` z `sklearn.metrics` pro výpočet úspěšnosti klasifikace (v zadání zmíněn `model_selection`, ale v Scikit-learn se metrika `accuracy_score` nachází v `metrics`).

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

## 2. Vytvoření syntetického datasetu

Parametry podle zadání:
- **Počet vzorků (`n_samples`):** 1000
- **Počet proměnných (`n_features`):** 10
- **Počet významných proměnných (`n_informative`):** 6

```python
X, y = make_classification(
    n_samples=1000,
    n_features=10,
    n_informative=6,
    random_state=42
)

print(f"Rozměry matice příznaků X: {X.shape}")
print(f"Rozměry vektoru cílové proměnné y: {y.shape}")
```

## 3. Rozdělení datasetu na trénovací a testovací sadu (70:30)

Poměr 70:30 specifikujeme parametrem `test_size=0.30`.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y, 
    test_size=0.30, 
    random_state=42
)

print(f"Trénovací sada: X_train = {X_train.shape}, y_train = {y_train.shape}")
print(f"Testovací sada:  X_test  = {X_test.shape}, y_test  = {y_test.shape}")
```

## 4. Vytvoření instance modelu k-NN (k = 4)

Zvolíme parametr `n_neighbors=4`.

```python
knn = KNeighborsClassifier(n_neighbors=4)
knn
```

## 5. Natrénování modelu a predikce na testovací sadě

```python
# Trénování modelu
knn.fit(X_train, y_train)

# Predikce na testovací sadě
y_pred = knn.predict(X_test)

print(f"Prvních 10 skutečných tříd:  {y_test[:10]}")
print(f"Prvních 10 predikcí modelu: {y_pred[:10]}")
```

## 6. Výpočet metriky přesnosti (Accuracy)

Výpočet hodnoty metriky Accuracy na základě predikcí (`y_pred`) a skutečných hodnot (`y_test`).

```python
accuracy = accuracy_score(y_test, y_pred)
print(f"Přesnost (Accuracy score): {accuracy:.4f} ({accuracy * 100:.2f} %)")

# Ověření pomocí metody .score() přímo na instanci modelu:
print(f"Kontrola přes knn.score:   {knn.score(X_test, y_test):.4f}")
```
